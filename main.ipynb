```python
import pandas as pd 
import numpy as np
import plotly.express as px
import os
import warnings
import re

warnings.filterwarnings("ignore")
os.listdir('data')
```

```python
data = pd.read_csv("data/products_asos.csv", on_bad_lines= 'skip')
data.sample(5)
```

```python
data.info()
```

```python
data['price'] = pd.to_numeric(data['price'], errors='coerce') #Convertimos a número el precio a número y obligamos a NaN si errores
data.dropna(subset=['price'], inplace= True) # Quitamos los NaNs

print("Data Shape: " + str(data.shape))
```

```python
# Problema a resolver: Analizar el precio para saber cuanto estan perdiendo por no tener productos en stockç

# Primero limpiamos la columna description para obtener la marca del producto usando la url para obtener la marca
def get_brand_from_url(url):
    match = re.search(r'asos\.com/([^/]+)/', url)

    if match:
        return match.group(1)

    return None

data["brand"] = data["url"].apply(get_brand_from_url)
```

```python
data.sample(5)
```

```python
# Limpiamos en nombre de la marca
data["brand"] = (
    data["brand"]
    .str.replace("-", " ")
    .str.title()
)
```

```python
# Poner en mínisculas todas los colores
def lower_string(text):
    return text.lower()

data['color'] = data['color'].apply(lower_string)
```

```python
# Separar la variable de size 
data["size_list"] = data["size"].str.split(",")

data = data.explode("size_list")

data.head(5)
```

```python
# Si size contiene Out of Stock, se crea una variable booleana
data["in_stock"] = ~data["size_list"].str.contains(
    "Out of stock",
    case=False,
    na=False
)
```

```python
# Contar nº de tallas por prenda
data["num_sizes"] = data["size"].str.count(",") + 1

# Contar las veces que se encuentra Out of Stock una talla de un producto
data["num_out_of_stock"] = data["size"].str.count("Out of stock")

# Definir un ratio de disponibilidad
data["availability_ratio"] = (
    data["num_sizes"] - data["num_out_of_stock"]
) / data["num_sizes"]
```

```python
data.head(5)
```

```python
data.info()
```

```python
# REALIZAR AGRUPACIONES Y PLOTS PARA VER LOS DATOS

brand_counts = (
    data.groupby("brand")
        .size()
        .sort_values(ascending=False)
        .reset_index()
)

fig = px.bar(
    brand_counts,
    x = 'brand',
    y = 0,
    color= 'brand',
    title= "Which brands have the most products?",
    template = 'simple_white'
)

fig.show()
```

```python
brand_summary = (
    data.groupby("brand")
        .agg(
            num_products=("sku", "count"),
            avg_price=("price", "mean"),
            median_price=("price", "median"),
            avg_availability=("availability_ratio", "mean")
        )
        .sort_values("num_products", ascending=False)
        .reset_index()
)

fig = px.bar(
    brand_summary,
    x = 'brand',
    y = 'avg_availability',
    color = "brand",
    title= "Average Availability of brands",
    template = 'simple_white'
    
)

fig.show()
```

```python
fig = px.scatter(
    brand_summary,
    x = 'median_price',
    y = 'avg_availability',
    color = 'brand',
    title= "Average Price of brands and their average availability",
    template = 'simple_white'
    
)
fig.show()
```

```python
brand_summary.head(10)
```

```python
category_summary = (
    data.groupby("category")
        .agg(
            products=("sku", "count"),
            avg_price=("price", "mean"),
            median_price=("price", "median")
        )
        .sort_values("products", ascending=False)
)
```

```python
availability_by_brand = (
    data.groupby("brand")
        .agg(
            avg_availability=("availability_ratio", "mean")
        )
        .sort_values("avg_availability")
        .reset_index()
)

fig = px.bar(
    availability_by_brand,
    x = 'brand',
    y = 'avg_availability',
    color = 'brand',
    title = 'Which brands are frequently sold out?',
    template = 'simple_white'
)
fig.show()
```

```python
price_by_brand = (
    data.groupby("brand")
        .agg(
            avg_price=("price", "mean"),
            median_price=("price", "median"),
            max_price=("price", "max")
        )
        .sort_values("avg_price", ascending=False)
        .reset_index()
)

fig = px.bar(
    price_by_brand,
    x = 'brand',
    y = 'avg_price',
    color = 'brand',
    title = 'Which brands are premium?',
    template = 'simple_white'
)
fig.show()
```

```python
top_colors = (
    data.groupby("color")
        .size()
        .sort_values(ascending=False)
        .reset_index()
)

top_colors.head(5)

fig = px.bar(
    top_colors,
    x = 'color',
    y = 0,
    color = "color",
    title = 'What colors are usually sold the most?',
)
fig.show()
```

```python
colors_price = data.groupby('color')['price'].mean().round(2).sort_values(ascending=False).reset_index()

fig = px.bar(
    colors_price,
    x = 'color',
    y = 'price',
    color = "color",
    title = 'What colors are usually sold with the highest price?',
)
fig.show()
```

```python
data_corr = data[[
    "price",
    "num_sizes",
    "availability_ratio"
]].corr(numeric_only=True)

# Do expensive products offer more sizes?
# Are expensive products more likely to sell out?

fig = px.imshow(data_corr)
fig.show()
```
